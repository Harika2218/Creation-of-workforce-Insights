# Phase 1: What is a Model?

An AI Model is a mathematical engine that has been trained to predict the next piece of information in a sequence.

Large Language Models (LLMs): These are models trained on massive amounts of text data (the internet, books, code, scientific papers).

Tokens: Models don't read words; they read "tokens" (chunks of characters).

The Transformer Architecture: This is the "brain structure" invented by Google in 2017 that allows models to understand the context of a word based on the words far away from it. This is why a model can remember the beginning of your Python script while writing the end.


#Phase:2 How Models are "Taught" (The Three Stages)
Not all models are created equal. Their skills depend on how they were raised:

Pre-training: The model reads the whole internet. It learns grammar, facts, and basic coding syntax.

Fine-Tuning (SFT): The model is specifically taught how to follow instructions. This is where a model learns, "When a user asks for Python, don't just talk about Python—write the code."

RLHF (Reinforcement Learning from Human Feedback): Humans rank the model’s answers. This makes the model more helpful, safer, and less likely to produce "hallucinations" (fake code).

#Phase 3: The Three Categories of Models
When you are working at a place like Infosys, you will encounter three types of models:

1. General Purpose Models (The "Polymaths")
Examples: GPT-4o, Claude 3.5 Sonnet, Gemini 1.5 Pro.
Pros: They are good at everything. They can write a Python script, then explain it in plain English, then write a business email about it.
Cons: They are expensive to run at a massive scale and can sometimes be "too wordy."

2. Code-Specific Models (The "Specialists")
Examples: DeepSeek-Coder, CodeLlama, StarCoder.
Pros: These models were trained on 80% code and 20% text. They understand logic, debugging, and syntax better than general models. They are often smaller and faster.
Cons: They might struggle with general conversation or complex creative writing.

3. Small/Local Models (The "Efficiency Experts")
Examples: Llama 3 (8B), Phi-3, Mistral.
Pros: These can run on your actual laptop without an internet connection. This is huge for Data Privacy (important for bank clients at Infosys).
Cons: They aren't as "smart" as the giant models.

#Phase 4: Models to Help with Your Specific Tasks
Now that we have the background, here are the specific models you should research for your Python and SQL work:

For Python Tasks (Logic & Scripting)
Claude 3.5 Sonnet: Currently the "King of Coding." It is incredibly good at Refactoring (taking messy code and making it clean) and following complex logic.

GPT-4o: The most reliable for Libraries. If you need to use specific Python libraries like Pandas for data or FastAPI for web, GPT-4o has seen the most documentation.
For SQL Tasks (Data & Databases)

SQLCoder (by Defog): A specialized model built specifically to turn "Natural Language" into "SQL Queries." It understands database schemas better than general models.

Gemini 1.5 Pro: Because of its Huge Context Window, you can upload your entire Database Schema (100+ table definitions) and ask, "Write a query to find the top-performing employee in the HR department." It won't forget the table names.
For Debugging & Code Review
DeepSeek-Coder-V2: An open-source model that is currently beating GPT-4 in many coding benchmarks. It is excellent at finding bugs in existing code.
