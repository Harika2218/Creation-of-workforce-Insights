```python
 #Prompt: You are a Senior Python Developer.
Context: I have a text file named server.log that contains hundreds of lines of logs. Each line follows this format: [TIMESTAMP] LEVEL - MESSAGE (e.g., [2023-10-27 10:15:02] ERROR - Database connection failed).
Instruction: Write a Python script that reads this file and counts how many times "ERROR", "WARNING", and "INFO" appear.
Constraint: Do not use any external libraries; use only standard Python libraries (like os or json). The script must handle the case where the file is missing without crashing.
Output: The final count should be printed to the console and also saved into a file called summary.json.

```

```python
import json
import os

def analyze_logs(input_file, output_file):
    # Initialize our counters
    report = {
        "INFO": 0,
        "WARNING": 0,
        "ERROR": 0
    }

    # Constraint Check: Handle missing file
    if not os.path.exists(input_file):
        print(f"Error: The file '{input_file}' was not found.")
        return

    try:
        with open(input_file, 'r') as file:
            for line in file:
                # Check for keywords in each line
                if "INFO" in line:
                    report["INFO"] += 1
                elif "WARNING" in line:
                    report["WARNING"] += 1
                elif "ERROR" in line:
                    report["ERROR"] += 1

        # Output 1: Print to console
        print("--- Log Analysis Summary ---")
        for level, count in report.items():
            print(f"{level}: {count}")

        # Output 2: Save to JSON
        with open(output_file, 'w') as json_file:
            json.dump(report, json_file, indent=4)
        print(f"\nSummary successfully saved to {output_file}")

    except Exception as e:
        print(f"An unexpected error occurred: {e}")

if __name__ == "__main__":
    # Specify the file names
    log_filename = "server.log"
    summary_filename = "summary.json"

    # Run the function
    analyze_logs(log_filename, summary_filename)
```
