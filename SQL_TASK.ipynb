```python
Role: You are a Senior Database Engineer.
Context: A company wants to manage employee records across different departments. They need to track employees, their departments, and salaries.
Instruction: Write SQL scripts to:

Create two tables: departments and employees with proper relationships.
Insert sample data.
Write queries to:
Find the highest paid employee in each department.
Calculate average salary per department.
List employees who earn more than the average salary of their department.
Constraint: Use PostgreSQL syntax. Include foreign keys, proper data types, and handle the case where tables might already exist. No external extensions.
Output: Provide clean, well-commented SQL code.
```

```python
-- =============================================
-- 1. Create Tables
-- =============================================
CREATE TABLE IF NOT EXISTS departments (
    dept_id SERIAL PRIMARY KEY,
    dept_name VARCHAR(100) NOT NULL UNIQUE,
    location VARCHAR(100)
);

CREATE TABLE IF NOT EXISTS employees (
    emp_id SERIAL PRIMARY KEY,
    emp_name VARCHAR(100) NOT NULL,
    dept_id INTEGER REFERENCES departments(dept_id),
    salary INTEGER NOT NULL,
    hire_date DATE DEFAULT CURRENT_DATE
);

-- =============================================
-- 2. Insert Sample Data
-- =============================================
INSERT INTO departments (dept_name, location) VALUES
    ('Engineering', 'Bangalore'),
    ('Sales', 'Mumbai'),
    ('HR', 'Delhi');

INSERT INTO employees (emp_name, dept_id, salary, hire_date) VALUES
    ('Amit Sharma', 1, 85000, '2022-03-15'),
    ('Priya Patel', 1, 92000, '2021-07-20'),
    ('Rahul Verma', 2, 65000, '2023-01-10'),
    ('Sneha Gupta', 2, 78000, '2022-09-05'),
    ('Vikram Singh', 3, 55000, '2023-05-12'),
    ('Ananya Rao', 1, 88000, '2020-11-30');

-- =============================================
-- 3. Query 1: Highest paid employee in each department
-- =============================================
SELECT
    d.dept_name,
    e.emp_name,
    e.salary
FROM employees e
JOIN departments d ON e.dept_id = d.dept_id
WHERE (e.dept_id, e.salary) IN (
    SELECT dept_id, MAX(salary)
    FROM employees
    GROUP BY dept_id
);

-- =============================================
-- 4. Query 2: Average salary per department
-- =============================================
SELECT
    d.dept_name,
    ROUND(AVG(e.salary), 2) AS avg_salary
FROM employees e
JOIN departments d ON e.dept_id = d.dept_id
GROUP BY d.dept_name
ORDER BY avg_salary DESC;

-- =============================================
-- 5. Query 3: Employees earning above department average
-- =============================================
SELECT
    e.emp_name,
    d.dept_name,
    e.salary,
    ROUND(avg_sal.avg_salary, 2) AS dept_avg
FROM employees e
JOIN departments d ON e.dept_id = d.dept_id
JOIN (
    SELECT dept_id, AVG(salary) AS avg_salary
    FROM employees
    GROUP BY dept_id
) avg_sal ON e.dept_id = avg_sal.dept_id
WHERE e.salary > avg_sal.avg_salary;
```
